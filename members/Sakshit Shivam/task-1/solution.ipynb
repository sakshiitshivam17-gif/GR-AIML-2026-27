# House Price Prediction using NumPy
This notebook contains my Task 1 mini-project. I used NumPy to analyse house sizes and prices and build a simple linear regression model.

## 1. Import NumPy and Create the Dataset

```python
import numpy as np

house_sizes = np.array([500, 800, 1000, 1200, 1500, 1800, 2000, 2200, 2500, 2800])
house_prices = np.array([15, 24, 30, 36, 45, 54, 60, 66, 75, 84])

print('House sizes:', house_sizes)
print('House prices (lakhs):', house_prices)
```

## 2. Explore the Data

```python
print('Number of houses:', len(house_sizes))
print('Mean size:', np.mean(house_sizes))
print('Mean price:', np.mean(house_prices))
print('Median price:', np.median(house_prices))
print('Minimum price:', np.min(house_prices))
print('Maximum price:', np.max(house_prices))
print('Price standard deviation:', np.std(house_prices))
```

## 3. Split Training and Testing Data

```python
X_train = house_sizes[:8]
y_train = house_prices[:8]
X_test = house_sizes[8:]
y_test = house_prices[8:]

print('Training sizes:', X_train)
print('Training prices:', y_train)
print('Testing sizes:', X_test)
print('Testing prices:', y_test)
```

## 4. Train the Linear Regression Model

```python
slope = np.sum((X_train - np.mean(X_train)) * (y_train - np.mean(y_train))) / np.sum((X_train - np.mean(X_train)) ** 2)
intercept = np.mean(y_train) - slope * np.mean(X_train)

print('Slope:', slope)
print('Intercept:', intercept)
print('Predicted price for 1300 sq ft:', slope * 1300 + intercept, 'lakhs')
```

## 5. Evaluate the Model

```python
predictions = slope * X_test + intercept
errors = y_test - predictions
mae = np.mean(np.abs(errors))
mse = np.mean(errors ** 2)

print('Actual prices:', y_test)
print('Predicted prices:', predictions)
print('MAE:', mae)
print('MSE:', mse)
```

## 6. Check the Effect of an Outlier

```python
prices_with_outlier = np.append(house_prices, 200)

print('Original mean:', np.mean(house_prices))
print('Mean with outlier:', np.mean(prices_with_outlier))
print('Original median:', np.median(house_prices))
print('Median with outlier:', np.median(prices_with_outlier))
```

## 7. Compare with a Baseline

```python
baseline_predictions = np.full(len(y_test), np.mean(y_train))
baseline_mae = np.mean(np.abs(y_test - baseline_predictions))
print('Baseline predictions:', baseline_predictions)
print('Baseline MAE:', baseline_mae)
print('Linear regression MAE:', mae)
```

## Conclusion
The model predicts house prices from house size using a simple linear relationship. The test error is zero on this small dataset, but real house prices depend on other factors too.
